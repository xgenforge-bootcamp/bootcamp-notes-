Human Neuron – Biological Inspiration for Deep Learning

A biological neuron processes information in the following way:

Dendrites receive signals from other neurons

The cell body (soma) combines and processes these signals

If the combined signal crosses a threshold, the neuron fires

The electrical impulse travels through the axon

The signal reaches axon terminals and is passed to the next neuron

This sequence can be summarized as:

Input → Combine → Decide → Output

Artificial neurons (perceptrons) are designed using the same idea.
They take multiple inputs, combine them using weights and bias, and decide the output using an activation function.

A real neuron takes inputs through dendrites, combines them in the cell body, and fires an output through the axon if the signal is strong enough.
Artificial neurons copy this behavior — they take inputs, combine them mathematically, and produce an output.”

## Intuition of Weights, Bias and Activation Function

When we talk about a single neuron in deep learning, three things make it “intelligent”:

- **Weights** → what to pay attention to  
- **Bias** → where to start from (baseline / threshold)  
- **Activation function** → how to decide the final output  

A neuron mathematically does this:

\[
z = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b
\]
\[
y = f(z)
\]

Where:
- \( x_i \) = inputs  
- \( w_i \) = weights  
- \( b \) = bias  
- \( f \) = activation function  
- \( y \) = output of the neuron  

---

### 1️⃣ Weights – “How important is this input?”

**Idea:**  
Weights are **importance scores** for each input.  
The model *learns* these weights during training.

- Large positive weight → this input strongly **increases** output  
- Large negative weight → this input strongly **decreases** output  
- Small weight → this input hardly affects the output  

We can think of a neuron as:
> “I will look at all inputs, but I will listen more to the ones with higher weights.”

#### Example 1 – House Price Prediction

Inputs:
- \( x_1 \) = size (sq. ft.)  
- \( x_2 \) = location score  
- \( x_3 \) = age of the house  

Suppose after training, the model learns:

- \( w_1 = 0.8 \) (size)  
- \( w_2 = 1.5 \) (location)  
- \( w_3 = -0.3 \) (age)  

Interpretation:
- Location is **most important** for price (1.5 is largest).  
- Size is important but less than location.  
- Age has a **negative weight** → older houses reduce the price.

So weights are telling the model:
> “Location matters a lot, size matters somewhat, and older age is bad for price.”

#### Example 2 – Email Spam Detection

Inputs:
- \( x_1 \) = “free money” phrase present (0 or 1)  
- \( x_2 \) = number of links in email  
- \( x_3 \) = email length  

Possible learned weights:
- \( w_1 = 2.0 \)  
- \( w_2 = 1.2 \)  
- \( w_3 = 0.1 \)  

Interpretation:
- If email contains “free money”, model leans heavily toward **spam**.  
- Many links also increase spam probability.  
- Length doesn’t matter much (weight close to 0).

#### Example 3 – Sentiment Detection

Inputs:
- \( x_1 \) = count of positive words (“great”, “awesome”)  
- \( x_2 \) = count of negative words (“bad”, “terrible”)  

Weights:
- \( w_1 = 1.0 \) (positive words)  
- \( w_2 = -1.3 \) (negative words)  

Interpretation:
- More positive words → increases sentiment score.  
- Negative words have stronger negative impact (–1.3), pulling score down more.

👉 **Summary for Weights:**  
Weights decide **how loudly each input speaks** in the neuron’s final decision.

---

### 2️⃣ Bias – “What is my starting point?”

**Idea:**  
Bias is a constant added to the weighted sum:

\[
z = w_1 x_1 + w_2 x_2 + \dots + b
\]

Bias lets the neuron **shift** its output up or down.  
It is like a “default tendency” or **baseline** before looking at inputs.

Without bias, the neuron’s decision boundary is forced to pass through 0, which is often too restrictive.

#### Example 1 – Exam Pass/Fail

Suppose a student needs **40 marks** to pass.

Simplified model:
- \( x \) = marks from questions  
- Weight \( w = 1 \)  

We want:
- If student gets 40 marks → just passes  
- If less than 40 → fails  

We can model this as:

\[
z = 1 \cdot x - 40
\]

Here, **bias b = -40**.  
So:
- If \( x = 40 \) → \( z = 0 \) (borderline)  
- If \( x > 40 \) → \( z > 0 \) → pass  
- If \( x < 40 \) → \( z < 0 \) → fail  

Bias (–40) sets the **pass threshold**.

#### Example 2 – Always Slightly Positive by Default

Imagine a product rating model which is slightly optimistic by default.

- Even if inputs are neutral, we want a small positive score.  

This can be done by using **a positive bias** b:

\[
z = w_1 x_1 + w_2 x_2 + b
\]

If all inputs are 0, then \( z = b \).  
So if \( b = 0.5 \), the model starts from 0.5 (slightly positive).

#### Example 3 – Hunger Analogy

Think about hunger level:

- Even if you haven’t seen food yet (inputs = 0), you may still feel **some hunger**.  
- That starting hunger level is like **bias**.  

Food appearance (inputs) increases hunger further, but baseline hunger was already there.

👉 **Summary for Bias:**  
Bias shifts the neuron’s decision boundary and gives it flexibility to learn patterns that **do not pass through zero**.

---

### 3️⃣ Activation Function – “How do I fire?”

**Idea:**  
Activation function decides **how the neuron responds** to the combined signal \( z \).

Without an activation function, the neuron is just a linear function → no matter how many layers we stack, the entire network behaves like **one big linear model**.

Activation introduces **non-linearity**, which allows the network to learn complex relationships.

\[
y = f(z)
\]

#### Common Activation Functions

1. **Step (Threshold) Activation – Intuition**

Old-style perceptron:
- If \( z > 0 \) → output 1  
- Else → output 0  

Good for:
- Yes / No decisions  
- But not good for modern deep learning (not differentiable).

2. **Sigmoid Activation**

\[
f(z) = \frac{1}{1 + e^{-z}}
\]

- Output is always between 0 and 1  
- Good for probabilities, e.g. spam vs not spam  

Examples:
- \( z = 0 \) → \( f(z) = 0.5 \) (uncertain)  
- \( z = 3 \) → \( f(z) \approx 0.95 \) (strong yes)  
- \( z = -3 \) → \( f(z) \approx 0.047 \) (strong no)

3. **ReLU (Rectified Linear Unit)**

\[
f(z) = \max(0, z)
\]

- If z is negative → output 0  
- If z is positive → output z  

Examples:
- \( z = -5 \) → output 0  
- \( z = 2.7 \) → output 2.7  

ReLU is simple and works very well in deep networks.

#### Example 1 – Neuron with Step-Like Behaviour

Imagine a neuron used to decide:

> “Is the temperature high enough to turn on the fan?”

- Inputs: temperature  
- If temperature > 30°C → output 1 (turn on)  
- Else → output 0 (keep off)  

This is a threshold activation.

#### Example 2 – Sigmoid for Probability

In logistic regression or a single output neuron:

- If sigmoid output = 0.9 → 90% chance of class 1  
- If sigmoid output = 0.1 → 10% chance of class 1 (so likely class 0)  

The activation function converts raw score z into **probability-like** output.

#### Example 3 – ReLU in Deep Networks

In a deep network layer, each neuron computes:

\[
z = w_1 x_1 + w_2 x_2 + ... + b
\]
\[
y = \max(0, z)
\]

- If neuron’s computed z is negative → we treat it as “inactive” (0).  
- If positive → neuron “fires” proportionally to how strong the signal is.  

This helps model sparse and efficient representations.

👉 **Summary for Activation:**  
Activation function is the neuron’s **firing rule**.  
It turns the raw score into a meaningful output and allows the network to learn highly non-linear decision boundaries.

---

### 🔚 Final Intuition Summary

For a single neuron:

\[
y = f(w_1x_1 + w_2x_2 + ... + b)
\]

- **Weights** decide *what* to focus on (importance of each input).  
- **Bias** decides *from where to start* (baseline / threshold).  
- **Activation function** decides *how to respond* (firing behaviour).

Together, they turn a simple mathematical formula into a powerful, learnable decision unit, which is the foundation of all deep learning models.


1. What is a Layer?

A layer is simply a collection of neurons stacked horizontally.

Each neuron inside the layer:

Receives the same input

Has different weights & bias

Learns a different pattern

Example:

Input → [Neuron1, Neuron2, Neuron3, ... NeuronN] → Output


If you see a layer with 128 neurons, it means the model is learning 128 different patterns at that step.

⭐ 2. How a Layer Works (Mathematical View)

Given:

Input vector X of size (n)

Weight matrix W of size (m × n)

Bias vector b of size (m)

The layer computes:

Layer Output = activation( W * X + b )


Where:

m = number of neurons

n = number of input features

Every neuron produces 1 value → entire layer produces a vector of size m.

⭐ 3. Why Multiple Layers? (Feature Hierarchy)

In classic ML, we had to manually create features:

N-grams

Lemmatization

POS tags

Sentiment lexicons

But Deep Learning automatically learns these patterns.

📌 Layer 1 → Low-level features

Learns simple word relationships:

presence of “good”, “bad”, “not”, etc.

📌 Layer 2 → Mid-level features

Learns combinations:

“not good”

“very bad”

“extremely helpful”

📌 Layer 3 → High-level meaning

Learns abstract semantic concepts:

sentence sentiment

intent

topic

This is called hierarchical feature learning.

⭐ 4. Intuition: Layers = Filters That Learn Meaning

Every layer transforms the input into a better representation.

Example analogy:

➡️ Layer 1: sees letters
➡️ Layer 2: sees words
➡️ Layer 3: sees sentences / meaning

The deeper the network, the richer the understanding.

⭐ 5. Layers in Real NLP Pipeline
Step 1 → Convert text to numbers (embedding)
Step 2 → Pass through Layer 1

Learns basic semantic signals.

Step 3 → Pass through Layer 2

Learns phrase-level patterns.

Step 4 → Pass through Layer 3

Learns sentence-level meaning.

Step 5 → Output layer

Gives prediction (positive, negative, category, etc.)

Example for Text Classification

Input sentence:
“The service was not good.”

Model learns:

Layer 1:

detects “not”

detects “good”

Layer 2:

learns phrase “not good”

Layer 3:

learns negative sentiment pattern

Final output:
➡️ Negative sentiment

1. What is a Feedforward Neural Network (FFNN)?

A feedforward neural network is the most basic neural network architecture, where information moves in one direction only:

Input → Hidden Layers → Output


There are no loops, no memory, no feedback connections.

This makes FFNN simple and fast.

2. Why FFNN Is Used in NLP

Text classification requires converting a sentence into a fixed-size vector, such as:

Bag of Words

TF-IDF

Word Embeddings

Sentence Embeddings

Once converted, this vector is fed into the FFNN.

Example vector (simplified):

[0.12, 0.00, 0.87, 0.33, ...]


The network then learns patterns in these numerical values.

Step 1: Text → Vector

Examples:

“This phone is amazing”

Step 2: Vector → Hidden Layer
h1 = activation( W1 · x + b1 )

“good”, “amazing”, “bad”, “not”

For binary classification:

ŷ = sigmoid( W2 · h1 + b2 )

## 1) What is a Feedforward Neural Network (FFNN)?

A Feedforward Neural Network is the simplest type of neural network where information flows in only one direction:

Input → Hidden Layer(s) → Output

There are **no loops** and no "memory".  
That means the network cannot remember past inputs while processing the current input.

In NLP, FFNN is mainly used for **simple classification tasks** where we convert text into a fixed-size numeric vector first
(e.g., BoW, TF-IDF, or an embedding), and then the network predicts the label.

Example tasks:
- Spam vs Not Spam
- Positive vs Negative sentiment
- Topic classification


## 2) How FFNN Works for Text (Step-by-Step)

FFNN cannot directly take raw text like "I love AI".
Text must be converted into numbers first.

### Step 1: Convert text into a numeric vector
Common options:
- Bag of Words (BoW): counts of words
- TF-IDF: weighted counts
- Embeddings: dense semantic vectors (Word2Vec, GloVe, or modern embeddings)

Result: a fixed-size vector like:
x = [0.2, 0.0, 0.9, 0.1, ...]

### Step 2: Pass this vector through hidden layers
Each hidden layer applies:
- a linear transformation (weights + bias)
- a non-linear activation function

This allows the model to learn patterns like:
- certain words strongly indicate spam
- some word combinations indicate negative sentiment

### Step 3: Output layer produces prediction
Depending on the task:
- Binary classification: sigmoid → probability (0 to 1)
- Multi-class classification: softmax → probabilities across classes

Example:
Output = 0.92 → "Spam"
Output = 0.12 → "Not Spam"


## 3) Why FFNN Works in NLP (Where it fits)

FFNN works well when:
✅ the text is short  
✅ the task is simple  
✅ order of words is not critical  
✅ you have good input features (BoW/TF-IDF/Embeddings)

Because FFNN learns patterns directly from the input features.

Example:
If "free", "winner", "click", "prize" appear often in spam emails,
FFNN learns this correlation and predicts spam.

So FFNN is:
- easy to build
- fast to train
- a good baseline model


## 4) Limitations of FFNN for NLP (Very Important)

FFNN cannot properly handle sequence or context because:

### 1) Cannot capture word order
Example:
"I love you" vs "You love I"
BoW/TF-IDF may look similar → FFNN may treat them similarly.

### 2) Cannot understand context (no memory)
Example:
"not good" vs "good"
FFNN may focus on "good" and miss the effect of "not" unless features capture it.

### 3) Works poorly for long text
Long documents create large sparse vectors (BoW/TF-IDF) and FFNN cannot track long-range meaning.

### 4) Not suitable for sequence modeling tasks
Tasks like:
- next word prediction
- translation
- summarization
need a model that understands sequence → FFNN fails.

👉 These limitations directly motivated RNN → LSTM → Transformers.


## 5) What is a Layer?

A layer is a group of neurons that processes input together.

- Input Layer: receives the input features (BoW / TF-IDF / Embedding vector)
- Hidden Layers: learn intermediate representations (patterns/features)
- Output Layer: produces final prediction (class probabilities)

Each neuron learns a different pattern from data.

Layers stacked together form a deep neural network.


## 6) Why Multiple Layers?

More layers allow the network to learn features in levels:

- Layer 1: learns simple patterns (basic word signals)
- Layer 2: learns combinations (important combinations of signals)
- Layer 3: learns higher-level meaning (more abstract patterns)

So deeper networks can model more complex relationships in the data.


## 7) Main Advantages of FFNN (vs pure ML)

✅ Less manual feature engineering (compared to classic ML)
✅ Learns useful feature representations automatically
✅ With embeddings, it can capture some semantic similarity better than BoW/TF-IDF

But remember:
FFNN still does not understand sequence.
It is a strong baseline, but not a sequence model.


### Final takeaway:
FFNN is good for simple text classification when text is converted to a fixed-size vector,
but it fails for sequences because it cannot learn word order or long-term context.
That’s why we move to RNN and LSTM next.

Traditional Feed Forward Neural Networks (FFNN) have limitations:

They process each input independently

They do not remember previous inputs

They fail for sequential data like text, speech, and time series

Language has order and context, so we need a model that can remember past information.

What is an RNN?

A Recurrent Neural Network (RNN) is a neural network designed for sequential data, where information from previous steps is carried forward using a hidden state (memory).

One-line definition

RNN reads inputs one by one and carries a small memory from previous steps.

Key Idea of RNN

At each time step t, an RNN receives:

xₜ → current input

hₜ₋₁ → previous hidden state (memory)

It combines both to produce:

hₜ → updated hidden state (new memory)

The hidden state acts as memory that flows through time.

## What is an RNN?

A Recurrent Neural Network (RNN) is a neural network designed for sequential data,
where information from previous steps is carried forward using a hidden state (memory).

### One-line definition
RNN reads inputs one by one and carries a small memory from previous steps.


## RNN Cell Concept (A block)

In the RNN diagram, the block **A** represents one RNN cell.
Unlike FFNN, this RNN cell has a loop.

That loop means the output of the hidden state is fed back into the cell,
so the network can "remember" information from the previous step.


## Key Idea of RNN

At each time step t, RNN receives:
- current input xₜ
- previous hidden state hₜ₋₁

It combines both to produce:
- new hidden state hₜ (updated memory)

So, the hidden state works as a memory that moves through time.

## How RNN works (Step-by-Step)

### Step 1: First input
- First input x₁ enters the network
- Initial hidden state h₀ is usually set to zero

### Step 2: Hidden state update
- RNN combines x₁ and h₀
- Produces new hidden state h₁

### Step 3: Memory flow
- Hidden state h₁ is passed to the next step
- This acts as memory for the sequence

### Step 4: Next time step
- At time step t, RNN takes:
  - current input xₜ
  - previous hidden state hₜ₋₁
- Produces updated memory hₜ

### Step 5: Sequence understanding
- By repeating this for all inputs,
  RNN builds contextual understanding of the sequence

### Important concept
- Hidden state acts like a moving summary of the sequence
- Information flows forward through time using the loop

- Language modeling
- Named Entity Recognition (NER)
- Time-series prediction


## Limitations of RNN

- Memory fades for long sequences
- Struggles with long-term dependencies
- Slow because processing is sequential

RNN understands sequence, but forgets long context → leads to LSTM.

## Why LSTM was introduced

Recurrent Neural Networks (RNNs) can process sequences, but they suffer from major problems:

- Memory fades for long sequences
- Vanishing gradient problem
- Unable to connect distant words effectively

### Example
"The movie that I watched last week was amazing"

RNN forgets the word "movie" by the time it reaches "amazing".

👉 LSTM was introduced to solve this **long-term dependency problem**.


## What is LSTM?

LSTM (Long Short-Term Memory) is a special type of RNN designed to
remember information for a long time.

LSTM is an RNN with a **controlled memory system** that decides:
- what to remember
- what to forget
- what to output


## Core Idea of LSTM

Unlike RNN (single memory), LSTM maintains **two states**:

- Cell State (Cₜ) → long-term memory
- Hidden State (hₜ) → short-term / working memory

Memory flow is controlled using **three gates**:

1. Forget Gate → removes unnecessary information
2. Input Gate → adds important new information
3. Output Gate → decides what information to expose

These gates act like **intelligent switches**.

## Diagram Notations

### xₜ (Input at time t)
Represents the current input to the LSTM cell.
In NLP, this is usually the embedding of the current word.

### hₜ₋₁ (Previous Hidden State)
Represents the short-term / working memory from the previous time step.
Helps the LSTM understand recent context.

### Cₜ₋₁ (Previous Cell State)
Represents the long-term memory carried across time.
Stores important information over long sequences.

### σ (Sigmoid Gate)
Outputs values between 0 and 1.
Acts like a switch to control information flow:
- 0 → block information
- 1 → allow information


## Gates in LSTM

### Forget Gate (σ)
Decides what information to remove from the previous cell state.
Prevents irrelevant or outdated memory from being carried forward.

### Input Gate (σ)
Decides how much new information should be added.
Controls writing new content into the cell state.

### tanh (Candidate Memory)
Generates new candidate information.
Values range from -1 to +1.
Works with the input gate to update memory.


## Step-by-Step Working of LSTM

### Step 1: Forget
The Forget Gate decides which information from the previous cell state Cₜ₋₁
is no longer useful and should be removed.

### Step 2: Add New Information
The Input Gate decides how much new information should be added,
and tanh generates candidate memory to update the cell state.

### Step 3: Update Memory
The old filtered memory and the new selected information are combined
to form the updated cell state Cₜ.

This cell state represents the long-term memory of the LSTM.

### Step 4: Output
The Output Gate decides what part of the updated memory should be exposed
as the hidden state hₜ.

The hidden state:
- is passed to the next LSTM cell
- is used as output for the current time step


## Key Takeaway

LSTM solves the long-term memory problem of RNN by introducing:
- a dedicated long-term memory (cell state)
- gates to control information flow

Simple intuition:
Forget → Add → Update → Output
